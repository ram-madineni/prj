```python
import tensorflow_datasets as tfds
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from sklearn.utils import resample
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# 1. Load the full EMNIST dataset ('byclass' = 62 classes: digits + upper + lower)
(ds_train, ds_test), ds_info = tfds.load(
    'emnist/byclass',
    split=['train', 'test'],
    as_supervised=True,
    with_info=True
)

# 2. Convert dataset to NumPy arrays
def convert_to_numpy(ds):
    images = []
    labels = []
    for image, label in tfds.as_numpy(ds):
        images.append(image)
        labels.append(label)  # No label offset needed for 'byclass'
    return np.array(images), np.array(labels)

X_train, y_train = convert_to_numpy(ds_train)
X_test, y_test = convert_to_numpy(ds_test)

# 3. Normalize and reshape
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0
X_train = X_train[..., np.newaxis]
X_test = X_test[..., np.newaxis]

# 4. Balance the training dataset
def balance_dataset(X, y):
    classes = np.unique(y)
    min_count = np.min([np.sum(y == c) for c in classes])
    X_balanced, y_balanced = [], []
    for c in classes:
        X_c = X[y == c]
        y_c = y[y == c]
        X_resampled, y_resampled = resample(X_c, y_c, n_samples=min_count, random_state=42)
        X_balanced.append(X_resampled)
        y_balanced.append(y_resampled)
    return np.concatenate(X_balanced), np.concatenate(y_balanced)

X_train_bal, y_train_bal = balance_dataset(X_train, y_train)

# 5. One-hot encode labels
num_classes = ds_info.features['label'].num_classes  # Should be 62
y_train_bal_oh = tf.keras.utils.to_categorical(y_train_bal, num_classes)
y_test_oh = tf.keras.utils.to_categorical(y_test, num_classes)

# 6. Build CNN model
def build_model():
    model = tf.keras.models.Sequential([
        tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
        tf.keras.layers.MaxPooling2D((2, 2)),
        tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
        tf.keras.layers.MaxPooling2D((2, 2)),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dropout(0.5),
        tf.keras.layers.Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])
    return model

model = build_model()

# 7. Train model
history = model.fit(
    X_train_bal, y_train_bal_oh,
    epochs=10,
    batch_size=128,
    validation_split=0.1
)

# 8. Evaluate model
test_loss, test_acc = model.evaluate(X_test, y_test_oh, verbose=0)
print(f"\nTest accuracy: {test_acc:.4f}")

# 9. Classification report
y_pred = np.argmax(model.predict(X_test), axis=1)
label_names = [ds_info.features['label'].int2str(i) for i in range(num_classes)]
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=label_names))

# 10. Confusion matrix
conf_mat = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(14, 12))
sns.heatmap(conf_mat, xticklabels=label_names, yticklabels=label_names, cmap='Blues', annot=False)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()

# 11. Visualize predictions
def visualize_predictions(X, y_true, y_pred, num_images=5):
    indices = np.random.choice(len(X), num_images, replace=False)
    plt.figure(figsize=(15, 3))
    for i, idx in enumerate(indices):
        plt.subplot(1, num_images, i + 1)
        plt.imshow(X[idx].squeeze(), cmap='gray')
        true_char = ds_info.features['label'].int2str(y_true[idx])
        pred_char = ds_info.features['label'].int2str(y_pred[idx])
        plt.title(f"True: {true_char}\nPred: {pred_char}")
        plt.axis("off")
    plt.show()

visualize_predictions(X_test, y_test, y_pred)
```

```python
# Write Streamlit app code
%%writefile app.py
import streamlit as st
import numpy as np
import tensorflow as tf
from PIL import Image
import cv2

# Load the trained model
# Load the model from the correct path in Google Drive
model = tf.keras.models.load_model('/content/drive/MyDrive/emnist_model.h5')

# Label names for EMNIST "byclass" (0–9, A–Z, a–z)
label_names = [chr(i) for i in range(48, 58)] + [chr(i) for i in range(65, 91)] + [chr(i) for i in range(97, 123)]

st.title("EMNIST Multi-Character Recognition")
st.write("Upload an image with multiple handwritten characters.")

uploaded_file = st.file_uploader("Choose an image...", type=["jpg", "jpeg", "png"])

def segment_characters(image):
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    _, binary = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV)

    contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    char_images = []
    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        if w * h > 50:
            char = binary[y:y+h, x:x+w]
            char = cv2.resize(char, (28, 28), interpolation=cv2.INTER_AREA)
            char = char.astype("float32") / 255.0
            char = 1.0 - char
            char = char.reshape(1, 28, 28, 1)
            char_images.append((x, char))

    char_images.sort(key=lambda x: x[0])
    return [char for _, char in char_images]

if uploaded_file:
    image = Image.open(uploaded_file).convert("RGB")
    img_np = np.array(image)

    st.image(image, caption="Uploaded Image", use_column_width=False)

    char_imgs = segment_characters(img_np)

    if not char_imgs:
        st.warning("No characters found.")
    else:
        st.subheader("Detected Characters:")
        predicted_string = ""
        # Use a spinner to indicate processing
        with st.spinner("Processing image..."):
            for char in char_imgs:
                pred = model.predict(char)
                label = label_names[np.argmax(pred)]
                predicted_string += label

        st.write(f"**Predicted Text:** {predicted_string}")
```
